```python
import pandas as pd
display(Markdown("# 5. Skenario akuisisi data "))

# ==========================================
# SKENARIO AKUISISI DATA
# ==========================================

skenario = {
    "Jenis Data": [
        "Teks",
        "Gambar",
        "Audio",
        "Video",
        "Dokumen",
        "Spasial"
    ],

    "Sumber Data": [
        "Pengurus sanggar",
        "Kegiatan sanggar",
        "Latihan tari",
        "Pertunjukan tari",
        "Administrasi sanggar",
        "Lokasi kegiatan"
    ],

    "Metode Akuisisi": [
        "Pendataan anggota dan jadwal latihan",
        "Pengambilan foto menggunakan kamera",
        "Perekaman musik gamelan",
        "Perekaman pertunjukan menggunakan kamera",
        "Pengumpulan dokumen digital",
        "Pengambilan koordinat GPS"
    ],

    "Frekuensi": [
        "Setiap bulan",
        "Setiap kegiatan",
        "Setiap latihan terpilih",
        "Setiap pertunjukan",
        "Setiap ada dokumen baru",
        "Setiap ada lokasi baru"
    ]
}

df_skenario = pd.DataFrame(skenario)
display(df_skenario)


# ==========================================
# ASUMSI JUMLAH DATA DAN UKURAN RATA-RATA
# ==========================================

print("\nASUMSI JUMLAH DATA DAN UKURAN RATA-RATA")

jenis_data = [
    "Teks", "Gambar", "Audio",
    "Video", "Dokumen", "Spasial"
]

jumlah_unit = [500, 1000, 200, 100, 300, 50]

ukuran_rata = [
    "5 KB", "3 MB", "5 MB",
    "150 MB", "2 MB", "10 KB"
]

# Konversi semua ukuran menjadi MB
ukuran_mb = [5/1024, 3, 5, 150, 2, 10/1024]

df_asumsi = pd.DataFrame({
    "Jenis Data": jenis_data,
    "Jumlah Unit": jumlah_unit,
    "Ukuran Rata-rata": ukuran_rata
})

display(df_asumsi)


# ==========================================
# MENGHITUNG STORAGE BULAN PERTAMA
# ==========================================

storage_awal = sum(
    jumlah * ukuran
    for jumlah, ukuran in zip(jumlah_unit, ukuran_mb)
)

print(f"\nStorage Bulan Pertama: {storage_awal:.2f} MB")


# ==========================================
# ESTIMASI PERTUMBUHAN 10% SELAMA 12 BULAN
# ==========================================

print("\nESTIMASI STORAGE SELAMA 12 BULAN")

pertumbuhan = 0.10
akumulasi = 0
hasil = []

for bulan in range(1, 13):

    storage_bulanan = storage_awal * (1 + pertumbuhan) ** (bulan - 1)

    akumulasi += storage_bulanan

    hasil.append({
        "Bulan": bulan,
        "Data Baru (MB)": round(storage_bulanan, 2),
        "Akumulasi Storage (MB)": round(akumulasi, 2)
    })

df_hasil = pd.DataFrame(hasil)

display(df_hasil)


# ==========================================
# HASIL AKHIR
# ==========================================

print("\nHASIL ESTIMASI PENYIMPANAN")

print(f"Storage Bulan Pertama : {storage_awal:.2f} MB")
print(f"Pertumbuhan Per Bulan : {pertumbuhan * 100:.0f}%")
print(f"Total Storage 1 Tahun : {akumulasi:.2f} MB")
print(f"Total Storage 1 Tahun : {akumulasi / 1024:.2f} GB")
```

```python
import pandas as pd
display(Markdown("# 6. Estimasi total storage + cadangan 30% "))
# Data asumsi penyimpanan
data_storage = {
    "Jenis Data": [
        "Data Teks",
        "Data Gambar",
        "Data Audio",
        "Data Video",
        "Data Dokumen",
        "Data Spasial"
    ],
    "Jumlah Unit": [500, 1000, 200, 100, 300, 50],
    "Ukuran Rata-rata (MB)": [
        5 / 1024,
        3,
        5,
        150,
        2,
        10 / 1024
    ]
}

# Membuat DataFrame
df_storage = pd.DataFrame(data_storage)

# Menghitung total setiap jenis data
df_storage["Total Storage (MB)"] = (
    df_storage["Jumlah Unit"] *
    df_storage["Ukuran Rata-rata (MB)"]
)

# Menampilkan tabel
display(df_storage.round(2))

# Menghitung total keseluruhan
total_storage = df_storage["Total Storage (MB)"].sum()

# Menghitung cadangan 30%
cadangan = total_storage * 0.30

# Menghitung total akhir
total_akhir = total_storage + cadangan

print("\nHASIL ESTIMASI PENYIMPANAN")
print(f"Total Storage : {total_storage:.2f} MB")
print(f"Cadangan 30%  : {cadangan:.2f} MB")
print(f"Total Akhir   : {total_akhir:.2f} MB")
print(f"Total dalam GB: {total_akhir / 1024:.2f} GB")
```

# 7. Mengapa Multimedia Sebaiknya Tidak Langsung Disimpan sebagai BLOB?

BLOB (Binary Large Object) adalah tipe data yang dapat
menyimpan data biner, seperti gambar, audio, dan video,
secara langsung di dalam database.

Penyimpanan BLOB memiliki kelebihan, seperti pengelolaan file
dan metadata dalam satu sistem serta dukungan transaksi
database. Namun, file multimedia berukuran besar dapat
meningkatkan ukuran database, kebutuhan backup, dan beban
pemulihan data. Dampaknya bergantung pada DBMS dan arsitektur
yang digunakan.

Untuk arsip digital sanggar tari, file gambar, audio, dan video
dapat disimpan di file system atau object storage. Database
relasional menyimpan metadata seperti nama file, format,
ukuran, tanggal unggah, dan lokasi penyimpanan file.

Pendekatan ini memisahkan data terstruktur dari file multimedia
sehingga pengelolaan file besar dapat dilakukan secara terpisah.
Namun, backup, keamanan akses, dan hubungan antara metadata
serta file tetap harus dikelola dengan baik.

BLOB tetap dapat menjadi pilihan yang sesuai untuk file kecil
atau kebutuhan aplikasi tertentu. Jadi, pemilihan metode
penyimpanan perlu mempertimbangkan ukuran file, kebutuhan
akses, backup, keamanan, dan karakteristik aplikasi.
